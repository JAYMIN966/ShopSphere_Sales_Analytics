```python
import pandas as pd
```

```python
df = pd.read_csv("data/Sample_Superstore.csv")
df.head()
```

```python
df.info()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='mixed', dayfirst=False)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='mixed', dayfirst=False)

# Standardize to MM-DD-YYYY (zero-padded, dash-separated)
df['Order Date'] = df['Order Date'].dt.strftime('%m-%d-%Y')
df['Ship Date'] = df['Ship Date'].dt.strftime('%m-%d-%Y')
```

```python
print(df[['Order Date', 'Ship Date']].head(10))
```

```python
df.to_csv('Sample_Superstore_cleaned.csv', index=False)
print("\nSaved as Sample_Superstore_cleaned.csv")
```

```python
df.columns
```

```python
def clean_column_name(col):
    col = col[0].lower() + col[1:]      # lowercase first letter only
    col = col.replace(' ', '_')          # replace spaces with underscore
    col = col.replace('-', '_')          # also handle the dash in 'Sub-Category'
    return col

df.columns = [clean_column_name(c) for c in df.columns]
```

```python
df.to_csv('Sample_Superstore_cleaned.csv', index=False)
print("\nSaved as Sample_Superstore_cleaned.csv")
```

```python
# 1. Load with CORRECT encoding
print("Before fixing:")
print("Rows with non-ASCII in product_Name:", df['product_Name'].apply(lambda x: not str(x).isascii()).sum())
print("Rows with leading/trailing whitespace:", df['product_Name'].apply(lambda x: str(x) != str(x).strip()).sum())
print("Rows with bad customer_ID casing:", (~df['customer_ID'].str.match(r'^[A-Z]{2}-\d{5}$')).sum())
```

```python
df.head()
```

```python
# 2. Fix non-breaking spaces (\xa0) in product_Name
df['product_Name'] = df['product_Name'].str.replace('\xa0', ' ', regex=False)
```

```python
# 3. Strip leading/trailing whitespace in product_Name
df['product_Name'] = df['product_Name'].str.strip()
```

```python
# 4. collapse any accidental double spaces created by the above fixes
df['product_Name'] = df['product_Name'].str.replace(r'\s+', ' ', regex=True)
```

```python
# 5. Fix customer_ID casing (standardize to uppercase)
df['customer_ID'] = df['customer_ID'].str.upper()
```

```python
# 6. Standardize column names to full lowercase snake_case
df.columns = [c.replace('-', '_').replace(' ', '_').lower() for c in df.columns]
```

```python
print("\nAfter fixing:")
print("Rows with non-ASCII in product_name:", df['product_name'].apply(lambda x: not str(x).isascii()).sum())
print("Rows with leading/trailing whitespace:", df['product_name'].apply(lambda x: str(x) != str(x).strip()).sum())
print("Rows with bad customer_id casing:", (~df['customer_id'].str.match(r'^[A-Z]{2}-\d{5}$')).sum())
print("\nNew columns:", df.columns.tolist())
```
